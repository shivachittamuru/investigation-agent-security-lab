# Microsoft Entra Identity for AI Agents

## Learning Objective

Understand how Entra authenticates users and workloads, how agents obtain access to downstream resources, and why valid authentication does not imply permission to access business data.

**Core principle: Identity establishes who is acting; authorization determines what they can do.**

## Key Concepts

- **Human identity:** The employee who initiates a task.
- **Agent identity:** The identity representing an AI agent.
- **Managed identity:** An Azure-managed workload authentication mechanism that avoids manually storing credentials.
- **Delegated access (OBO):** Access to a downstream API on behalf of a signed-in user.
- **Application-only access:** Access using the application's or agent's own permissions.
- **Access token:** A token issued for a specific resource and permission context.

## Experiment

Acquire a real Entra-issued access token using the Azure CLI credential, inspect selected claims, and understand why that token cannot automatically authorize access to our investigation API.

## Expected Outcome

Understand the difference between authentication, token audience, scopes/roles, delegated permissions, agent identity, and application-specific resource authorization.

**Key takeaway: A valid token is not universal permission.**
